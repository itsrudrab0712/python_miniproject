```python
import cv2
import numpy
img=cv2.imread("lionbw.jpg",cv2.IMREAD_GRAYSCALE)
print(img.shape)
print(img)
rows=img.shape[0]
cols=img.shape[1]
for i in range(rows):
  for j in range(cols):
    if img[i,j]<120:
      img[i,j]=255
    elif img[i,j]>=120:
      img[i,j]=0
cv2.imwrite("lionbwedited.jpg",img)
```

```python
import cv2
img=cv2.imread("panda1.png")
print(img.shape)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
height=img.shape[0]
width=img.shape[1]
channels=img.shape[2]

for i in range(height):
  for j in range(width):
    pixel=img[i,j]
    if pixel[0]==0 or pixel[1]==255 or pixel[2]==0:
      img[i,j]=[255,255,255]
img=cv2.cvtColor(img,cv2.COLOR_RGB2BGR)
cv2.imwrite("Panda2.png",img)
```
